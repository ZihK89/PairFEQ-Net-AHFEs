```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
mols = [
    'mol_242480', 'mol_1396156', 'mol_2725802', 'mol_3034976', 'mol_3047394',
    'mol_4463913', 'mol_5003962', 'mol_6239320', 'mol_6338073', 'mol_7378987'
]
raw_energy_errors = ['0.06','0.02','0.09',' 0.04','0.23','0.16','0.03','0.01','0.20','0.11']
raw_charges_errors = ['0.0015','0.0019','0.0020',' 0.0009','0.0033','0.0036','0.0010','0.0006','0.0032','0.0015 ']

energy_errors = [float(e.strip()) for e in raw_energy_errors]
charges_errors = [float(c.strip()) for c in raw_charges_errors]
```

```python
fig, ax1 = plt.subplots(figsize=(9, 4))
x = np.arange(len(mols))
width = 0.30

ax1.bar(x - width/2, energy_errors, width, label='Energy Error (kcal/mol)', 
        edgecolor='black', linewidth=1.5, color='#b95a58')
ax1.set_ylabel('Energy Error (kcal/mol)', fontweight='bold', color='#b95a58')
ax1.set_ylim(0, 0.5)

ax1.set_xticks(x)
ax1.set_xticklabels(mols, rotation=45, ha='right', fontsize=10)

ax2 = ax1.twinx()
ax2.bar(x + width/2, charges_errors, width, label='Charges Error (e)', 
        edgecolor='black', linewidth=1.5, color='#86b5a1')
ax2.set_ylabel('Charges Error (e)', fontweight='bold',color='#86b5a1')
ax2.set_ylim(0, 0.0156)


lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

fig.tight_layout()
plt.title('PairF-Net Achieves QM Accuracy in Predicting Molecular Potential Energy and Charges')

# plt.savefig('mlp_errors.svg', dpi=300)
plt.show()
```
